```python
import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.model_selection import train_test_split

pd.set_option("display.max_columns", None)

print("Libraries loaded")
```

```python
notebook_dir = Path.cwd()

candidate_paths = [
    notebook_dir / "data/raw/MBA.csv",
    notebook_dir.parent / "data/raw/MBA.csv",
]

data_path = next((path for path in candidate_paths if path.exists()), None)

if data_path is None:
    raise FileNotFoundError(
        f"Could not find MBA.csv. Checked: {', '.join(str(path) for path in candidate_paths)}"
    )

df = pd.read_csv(data_path)

print("Shape:", df.shape)

df.head()
```

```python
df["admission_clean"] = df["admission"].fillna("Deny")

df["admission_clean"].value_counts()
```

```python
df["target"] = np.where(
    df["admission_clean"] == "Admit",
    1,
    0
)

df["target"].value_counts()
```

```python
columns_to_drop = [
    "application_id",
    "race",
    "gender",
    "admission",
    "admission_clean"
]

df = df.drop(columns=columns_to_drop)

print(df.columns)
```

```python
target_distribution = (
    df["target"]
    .value_counts(normalize=True)
    .mul(100)
)

print(target_distribution)
```

```python
for col in df.columns:
    print("\n")
    print(col)
    print(df[col].nunique())
```
